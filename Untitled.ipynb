```python
from mftool import Mftool
import pandas as pd
import time
import requests
from requests.exceptions import ChunkedEncodingError

# Initialize mftool
mf = Mftool()

# Get all mutual fund schemes
all_schemes = mf.get_scheme_codes()

# Filter schemes by keywords
target_keywords = ["Large Cap", "Mid Cap", "Small Cap", "Flexi Cap", "Bluechip"]
target_schemes = {code: name for code, name in all_schemes.items() if any(keyword in name for keyword in target_keywords)}

print(f"Found {len(target_schemes)} mutual funds matching target categories.")

# Collect NAV data for target funds
all_fund_data = []

for scheme_code, scheme_name in target_schemes.items():
    print(f"Fetching data for {scheme_name}...")
    retries = 3  # Number of retries
    for attempt in range(retries):
        try:
            nav_data = mf.get_scheme_historical_nav(scheme_code)
            if 'data' in nav_data and len(nav_data['data']) > 0:
                nav_df = pd.DataFrame(nav_data['data'])
                nav_df['Fund Name'] = scheme_name
                all_fund_data.append(nav_df)
                break  # Exit retry loop if successful
            else:
                print(f"Failed to fetch data for {scheme_name}.")
                break
        except ChunkedEncodingError:
            print(f"ChunkedEncodingError. Retrying {scheme_name}... Attempt {attempt + 1}")
            time.sleep(5)  # Wait before retrying

# Combine all data into a single DataFrame
if all_fund_data:
    combined_df = pd.concat(all_fund_data, ignore_index=True)
    combined_df['date'] = pd.to_datetime(combined_df['date'], format='%d-%m-%Y')
    combined_df.to_csv("target_category_fund_nav_data.csv", index=False)
    print("Data collection complete. Saved to target_category_fund_nav_data.csv.")
else:
    print("No data fetched for target category funds.")
```

```python
from statsmodels.tsa.stattools import adfuller
def check_stationarity(series):
    result = adfuller(series.dropna())
    print('ADF Statistic:', result[0])
    print('p-value:', result[1])
    if result[1] <= 0.05:
        print("Series is stationary")
    else:
        print("Series is not stationary")
# Checking stationarity of the 'nav' column in the train dataset
check_stationarity(train['nav'][-5000:])  # Checking last 5000 rows
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout

# Build the LSTM model
model = Sequential()

# LSTM layer with 50 units, input shape corresponds to the number of time steps and features
model.add(LSTM(units=50, return_sequences=True, input_shape=(X_train.shape[1], X_train.shape[2])))
model.add(Dropout(0.2))  # Dropout for regularization

# Another LSTM layer
model.add(LSTM(units=50, return_sequences=False))
model.add(Dropout(0.2))

# Output layer
model.add(Dense(units=1))

# Compile the model
model.compile(optimizer='adam', loss='mean_squared_error')
```
